```python
import psycopg2 as pg
```

```python
conexao = pg.connect(
    dbname="postgres",
    user="postgres",
    password='postgres',
    host='localhost',
    port='5432'
)
```

```python
cursor = conexao.cursor()
print('conectado!')
```

```python
cursor.execute('SELECT * FROM plataforma')
plataformas = cursor.fetchall()
print(plataformas)
```

```python
conexao.rollback()
for plataforma in plataformas:
    print(plataforma)
```

```python
class Plataforma ():
    nome: str
    fabricante: str

    def __init__(self,nome: str, fabricante: str):
        self.nome = nome
        self.fabricante=fabricante

    def __str__(self):
        return f'Plataforma({self.nome}, fabricante{self.fabricante})'

    def __repr__(self):
        return f'Plataforma {self.nome}'
```

```python
plataformas_para_inserir: list[Plataforma] = [
    Plataforma ('Playstation 3', 'Sony'),
    Plataforma ('Playstation 2', 'Sony'),
    Plataforma ('Playstation 1', 'Sony'),
    Plataforma ('Xbox Original', 'Microsoft'),
    Plataforma ('Xbox 360', 'Microsoft'),
    Plataforma ('Xbox One', 'Microsoft'),
    Plataforma ('Xbox Series X', 'Microsoft'),
]

print(plataformas_para_inserir)
```

```python
try:
    cursor.execute(
        'INSERT INTO plataforma(nome, fabricante) VALUES(%s,%s);',
        ('Playstation 2', 'Sony')
    )
except pg.DatabaseError as e:
    print(e)
    conexao.rollback()
```

```python
for plataforma in plataformas_para_inserir:
    try:
        print(plataforma)
        cursor.execute(
            'INSERT INTO plataforma(nome, fabricante) VALUES(%s,%s);',
            (plataforma.nome,plataforma.fabricante)
        )
        conexao.commit()
    except pg.DatabaseError as e:
        print(e)
        conexao.rollback()
```
